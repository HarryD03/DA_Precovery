```python
def somb(x: Union[DA, float], y: Union[DA, float]) -> Union[DA, float]:
    """Sombrero function"""
    r = sqrt(x ** 2 + y ** 2)
    s = sin(r) / r
    return s

# DA representation
DA.init(10,2)

x0 = 2.0
y0 = 3.0
x = DA(1)
y = DA(2)

f = somb(x0 + x, y0 + y)

#print and compare
print(f"DA method: x\n{f}\n")
```

```python
def Finite_Differences(x0: float, y0: float) -> np.ndarray:
    h = 1e-3
    dfdx = (somb(x0 +h, y0) - somb(x0-h,y0))/(2*h)    #central finite differences for dx
    dfdy = (somb(x0, y0+h) - somb(x0,y0-h))/(2*h)     #central fintie differences for dy
    dfdxx = ( somb(x0 + 2.0*h, y0) - 2.0*somb(x0, y0) + somb(x0 - 2.0*h, y0))    / (2.0 * h)**2
    dfdyy = (
        somb(x0, y0 + 2.0 * h)
        - 2.0 * somb(x0, y0)
        + somb(x0, y0 - 2.0 * h)) / (4.0 * h * h)
    dfdxxx = (
        somb(x0 + 3.0 * h, y0)
        - 3.0 * somb(x0 + h, y0)
        + 3.0 * somb(x0 - h, y0)
        - somb(x0 - 3.0 * h, y0)) / (8.0 * h * h * h)
    return np.array([dfdx, dfdy, dfdxx, dfdyy,dfdxxx])

def DA_coeff(f: DA) -> np.ndarray:
    dfdx = f.deriv(1).cons()
    dfdy = f.deriv(2).cons()
    dfdxx = f.deriv(1).deriv(1).cons()
    dfdyy = f.deriv(2).deriv(2).cons()
    dfdxxx = f.deriv(1).deriv(1).deriv(1).cons()
    f_DA_coeff = np.array([dfdx, dfdy, dfdxx, dfdyy,dfdxxx])
    return f_DA_coeff


f_diff = Finite_Differences(x0, y0)             #Do finite differences
f_DA_coeff = DA_coeff(f)
err = (f_DA_coeff- f_diff)


print(f"DA coefficient:\n {f_DA_coeff}\n")
print(f"Finite difference coefficient:\n {f_diff}\n")
print(f"Error:\n {err}\n")
```

```python
def f_DA(x: DA) -> DA:                                      #Function for DA number - change if want another equation
    return x * x.sin() + x.cos()

def newton(f: DA, x0: float) -> float:                      #Example of automatic differention, not polynomial evaluation. So                                                               we do not create a taylor map and evaluate it
    DA.pushTO(1)

    tol = 1e-6
    err = 1
    i = 0
    while err > tol:
        x0 -= f.cons() / f.deriv(1).cons()
        err = abs(err - tol)
        i += 1
    DA.popTO()

    print(f"root x0: \n {x0}\n")
    print(f"Newton iterations:\n {i}\n")
    print(f"Error:\n {err}\n")
    return x0

x0 = 3                                                      #Expand around 3
x = x0 + DA(1)                                              #Set as DA
f = f_DA(x)                                                 #Create taylor polynomial of DA function

newton(f, x0)
```
