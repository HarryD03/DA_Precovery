```python
def sombDA(x: array) -> DA:                 #Where x is an array in DA
    norm_x = x.vnorm()
    return norm_x.sin() / norm_x

def sombFP(x: float) -> float:              #Where x is an array in floats
    norm_x = np.linalg.norm(x, ord=None)
    return math.sin(norm_x) / norm_x

DA.cons()
```
